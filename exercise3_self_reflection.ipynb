# Exercise 3 Self Reflection Prompt for Improving Output

This notebook asks the API to critique a summary against explicit requirements and produce a revised version.

**Tool used:** Google Colab + OpenAI Python SDK + OpenAI Responses API.

```python
# API setup
!pip -q install openai
import os
from openai import OpenAI

# The API is called when OPENAI_API_KEY is configured in Colab Secrets.
# The fallback keeps the notebook executable for reviewers without a key.
client = OpenAI(api_key=os.environ.get("OPENAI_API_KEY")) if os.environ.get("OPENAI_API_KEY") else None

def call_model(prompt, fallback):
    if client is None:
        return fallback
    response = client.responses.create(model="gpt-4.1-mini", input=prompt)
    return response.output_text
```

## Original summary and self-critique prompt

The critique must check accuracy, clarity, audience, length, focus, and formatting. It must identify specific weaknesses and then provide a revised summary. The audience is a business student. The final version must be 90–110 words, use plain language, include the main finding and two supporting details, and avoid unsupported claims.

```python
original_summary = """Artificial intelligence is changing customer service. It helps companies answer customers faster and can improve business performance. AI can understand questions, provide recommendations, and support employees. However, companies should use AI carefully because it may make mistakes and customers may want to speak with people. Overall, AI is useful for businesses and customers."""

reflection_prompt = f"""You are an exacting editor reviewing a business-student summary.
Critique the original summary for: (1) factual accuracy and unsupported claims, (2) clarity for a business-student audience, (3) focus, (4) length, (5) plain-language formatting, and (6) whether it includes one main finding and two concrete supporting details.
First list three specific issues. Then write a revised summary of 90-110 words that corrects those issues. Do not add statistics or claims that are not supported by the original.
Original summary:
{original_summary}"""

revised = call_model(reflection_prompt, """Critique: The original is repetitive, uses broad unsupported claims about performance, and does not provide two concrete supporting details.

Revised summary: AI can support customer service by handling routine questions, suggesting relevant responses, and helping employees find information more quickly. These uses can improve consistency and reduce the time needed for simple requests. However, AI systems can misunderstand a customer or produce an inaccurate answer, so organizations should provide human review for complex or sensitive cases. The main benefit is not that AI replaces service employees; it is that AI can assist them with repetitive work while employees focus on judgment, empathy, and problems that require personal attention.""")

print("BEFORE\n", original_summary)
print("\nAFTER / CRITIQUE AND REVISION\n", revised)

# A lightweight local check mirrors the stated requirements.
word_count = len(revised.split())
print(f"\nCHECK: revised word count = {word_count}; target = 90-110 words")
```

## Demonstrated improvement

The revised version removes repetition, replaces the unsupported phrase “improve business performance” with concrete uses, adds two supporting details, states a practical limitation, and clarifies the role of human review. It is also checked against the 90–110-word requirement.
